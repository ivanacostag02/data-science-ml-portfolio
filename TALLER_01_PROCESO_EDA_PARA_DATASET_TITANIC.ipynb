Conversión a categóricas:

Pclass (clase del boleto) y Survived (supervivencia) se convierten a tipo category aunque sean numéricas, ya que representan clases/categorías discretas.

Label Encoding:

Para Sex y Embarked (texto), aplicamos LabelEncoder que asigna:

Sex: Male → 0, Female → 1

Embarked: C → 0, Q → 1, S → 2

Incluye manejo básico de valores faltantes (pd.NA).

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Convertir variables numéricas que son códigos a categóricas
numeric_codes = ['Pclass', 'Survived']

for col in numeric_codes:
    df_titanic[col] = df_titanic[col].astype('category')
```

```python
# 2. Aplicar Label Encoding a variables categóricas de texto
# Sex: Male → 0, Female → 1
# Embarked: C → 0, Q → 1, S → 2

text_categories = ['Sex', 'Embarked']

obj_le = LabelEncoder()

for col in text_categories:
    # Rellenar NaN temporalmente para el encoding
    df_titanic[col] = df_titanic[col].fillna('Missing')

    df_titanic[col] = obj_le.fit_transform(df_titanic[col])

    # Opcional: revertir fillna si se requiere
    # df_titanic[col] = df_titanic[col].replace(obj_le.transform(['Missing'])[0], pd.NA)

# Verificar cambios
print("\nTipos de datos después de las transformaciones:")
print(df_titanic.dtypes)

# Mostrar ejemplo de valores codificados
print("\nEjemplo de valores codificados:")
print(df_titanic[['Sex', 'Embarked']].head())
```
